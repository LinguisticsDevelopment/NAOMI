```python
s = "the boy is tall and fat".split(' ')
p = ['DET', 'NOUN', 'VERB', 'DET', 'COORD', 'DET']
c = [False]*len(s)
n = len(s)
s, p, c
```

```python
# parser state
pos = 0
```

```python
def get_unconsumed(d, part, curr):
    _pos = curr+d
    if _pos < 0 or _pos > n:
        return -1

    while c[_pos]:
        _pos += d
    
    if p[_pos] != part: #Check if this is right part of speech
        return -1       #Return -1 (out of index) if wrong

    return _pos
```

```python
def get_all(d, part):
    global pos
    words = []
    curr = get_unconsumed(d, part, pos)
    while curr != -1: #while getUnconsumed != -1, return getUnconsumed
        words.append(curr)
        curr = get_unconsumed(d, part, curr)
    
    return words

# s = "the boy is ugly, tall and fat".split(' ')
# p = ['DET', 'NOUN', 'VERB', 'DET', 'DET', 'COORD', 'DET']
# c = [False, False, False, False, False, False, False]
# n = len(s)
# pos = 5

# get_all(-1, 'DET')
```
